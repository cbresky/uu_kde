```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data\heart_failure_clinical_records_dataset.csv")
df.head()
```

## a. Kernel Density Estimate of ejection_fraction

```python
sns.kdeplot(data=df, x='ejection_fraction')
plt.show()
```

## b. Use Pandas' df.sample(frac=1.0, replace=True) to resample the data 15 times with replacement, plotting the kernel density estimtes for each sample.

```python
for i in range(15):
    resample = df.sample(frac=1.0, replace=True)
    sns.kdeplot(data=resample, x='ejection_fraction')

plt.title("KDE estimates for 15 resamples\n")
```

```python
# put original KDE plot on top to better compare 
for i in range(15):
    resample = df.sample(frac=1.0, replace=True)
    sns.kdeplot(data=resample, x='ejection_fraction', alpha=0.5)
sns.kdeplot(data=df, x='ejection_fraction', color='black', linewidth=2)

plt.title("Original (black) vs. 15 resampled KDEs\n")
plt.show()
```

## c. For what values of ejection_fraction is the original plot reliable? For which values is there noticeable variation in your plots of the resampled data?

- Looking for where the original plot is reliable (i.e., the resampled plots look similar), it looks like all of the resampled plots agree on the location of the peak density. It also looks like there seems to be a lot of overlap in the far tails as well, for values 70+ and for values below ~10 of ejection_fraction.
- It looks like there is a lot of variation especially around an ejection fraction of ~55-60 near the smaller hump in the density. There's also some variation between the resampled curves from ~25-30 ejection_fraction values.
